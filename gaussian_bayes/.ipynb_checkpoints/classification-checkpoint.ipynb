```python
from __future__ import division
import numpy as np
import pandas as pd
import sys
from scipy.stats import multivariate_normal

X = np.genfromtxt("X_train.csv", delimiter=",")
y = np.genfromtxt("y_train.csv", delimiter=",")
```

```python
# Train-test split

from sklearn.model_selection import train_test_split

# Split into 80% training and 20% testing
def splitTrain_test(X_train, y_train):
    X_train_split, X_test_split, y_train_split, y_test_split = train_test_split(
    X_train, y_train, test_size=0.2, random_state=42)

    return X_train_split, X_test_split, y_train_split, y_test_split

    
```

```python
# Train

def fit(X_train, y_train, k):
    
    classes = np.unique(y_train)
    n_classes = len(classes)
    n_features = X_train.shape[1]


    prior = np.zeros(n_classes)
    cov = np.zeros((n_classes, n_features, n_features))
    mean = np.zeros((n_classes, n_features))
    cls_length = np.zeros(n_classes)

    eps = 1e-6
    
    for i in range(n_classes):

        # class specific priors
        cls_prior = np.sum(y_train == i)/y_train.shape[0]
        prior[i] = cls_prior

        # class specific means
        cls_mean = np.mean(X_train[y_train == i], axis=0)
        mean[i] = cls_mean
        
        cls_data = X_train[y_train == i]
        cls_length[i] = cls_data.shape[0]
        cls_centered_data = cls_data - cls_mean
        
        # class specific covariance matrices
        cls_covMatrix = (cls_centered_data.T @ cls_centered_data)/(cls_data.shape[0]-1)
        
        cls_covMatrix += eps*np.eye(n_features)  # matrix regularization. only adding small values to the diagonal to avoid singular cov matrices
        cov[i] = cls_covMatrix


    return prior, cov, mean, n_classes, classes




def pluginClassifier(X_test, prior, cov, mean, n_classes):

    # Vectorized: compute log-likelihood for ALL samples at once per class
    n_samples = X_test.shape[0]
    scores = np.zeros((n_samples, n_classes))

    for c in range(n_classes):
        scores[:, c] = multivariate_normal.logpdf(
            X_test, mean=mean[c], cov=cov[c], allow_singular=True
        ) + np.log(prior[c])

    return np.argmax(scores, axis=1)

```

```python
# Test
# collect splitted data
X_np = X
y_np = y
X_train, X_test, y_train, y_test = splitTrain_test(X_np, y_np)


# fit
prior, cov, mean, n_classes, classes = fit(X_train, y_train)

# predict
pred_label_indexes = pluginClassifier(X_test, prior, cov, mean, n_classes)
predicted_labels = classes[pred_label_indexes]


print("Predicted label vector:\n", predicted_labels, predicted_labels.shape)

accuracy = np.mean(predicted_labels == y_test)
print(f"Accuracy: {accuracy:.4f}")
```

```python
#---------------------------------------------------------
#    USING K-FOLD CROSS VALIDATION to boost accuracy
#---------------------------------------------------------


# fit
from sklearn.model_selection import KFold
from sklearn.metrics import accuracy_score


def fit_cv(X_train, y_train, k=10):

    classes = np.unique(y_train)

    n_classes = len(classes)
    n_features = X_train.shape[1]

    kf = KFold(n_splits=k, shuffle=True, random_state=42)

    # storing all folds serially
    prior = []
    cov = []
    mean = []

    best_accuracy = -1
    best_fold = -1

    eps = 1e-6

    for fold, (train_idx, val_idx) in enumerate(kf.split(X_train)):

        X_tr = X_train[train_idx]    # k-1 train folds
        y_tr = y_train[train_idx]

        X_val = X_train[val_idx]     # 1 validation fold
        y_val = y_train[val_idx]

        fold_prior = np.zeros(n_classes)
        fold_cov = np.zeros((n_classes, n_features, n_features))
        fold_mean = np.zeros((n_classes, n_features))

        # train QDA for current fold
        for i, cls in enumerate(classes):

            cls_data = X_tr[y_tr == cls]

            # prior
            fold_prior[i] = cls_data.shape[0] / X_tr.shape[0]

            # mean
            cls_mean = np.mean(cls_data, axis=0)
            fold_mean[i] = cls_mean

            # covariance
            cls_centered = cls_data - cls_mean
            cls_cov = (cls_centered.T @ cls_centered) / (cls_data.shape[0] - 1)
            cls_cov += eps * np.eye(n_features)

            fold_cov[i] = cls_cov

        # store current fold params
        prior.append(fold_prior)
        cov.append(fold_cov)
        mean.append(fold_mean)

        # validation prediction
        predictions = []

        for x in X_val:

            scores = []

            for i in range(n_classes):

                score = multivariate_normal.logpdf(
                    x, mean=fold_mean[i], 
                    cov=fold_cov[i], 
                    allow_singular=True
                ) + np.log(fold_prior[i])

                scores.append(score)

            predictions.append(classes[np.argmax(scores)])

        predictions = np.array(predictions)

        acc = accuracy_score(y_val, predictions)

        print(f"Fold {fold + 1} Accuracy: {acc:.4f}")

        if acc > best_accuracy:
            best_accuracy = acc
            best_fold = fold

    print(f"\nBest Fold: {best_fold + 1}")
    print(f"Best Accuracy: {best_accuracy:.4f}")

    # trim arrays to only best fold
    prior = np.array(prior)[best_fold]
    cov = np.array(cov)[best_fold]
    mean = np.array(mean)[best_fold]

    return prior, cov, mean, n_classes, classes
```

```python
# Test
# collect splitted data
X_np = X
y_np = y
X_train, X_test, y_train, y_test = splitTrain_test(X_np, y_np)


# fit
prior, cov, mean, n_classes, classes = fit_cv(X_train, y_train)
```

```python
# predict
pred_label_indexes = pluginClassifier(X_test, prior, cov, mean, n_classes)
predicted_labels = classes[pred_label_indexes]


print("Predicted label vector:\n", predicted_labels, predicted_labels.shape)

accuracy = np.mean(predicted_labels == y_test)
print(f"Accuracy: {accuracy:.4f}")
```
